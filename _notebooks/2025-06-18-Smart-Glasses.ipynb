

# Smart Glasses — A Low-Cost Visual Aid (Prototype)

**Goal:** Build a low-cost wearable that helps people with partial vision loss (or elderly users) detect obstacles using an ultrasonic sensor and receive haptic or simple audio feedback through bone-conduction or a buzzer.

**Why I’m building this:** I want to make mobility safer and more independent for people in my community — especially older relatives who are prone to tripping or who struggle with navigating crowded indoor spaces. This project mixes hardware, embedded programming, and human-centered testing, and it’s inexpensive enough to prototype at home.

---

## Quick Overview
- **Core idea:** Mount a small ultrasonic rangefinder to a pair of glasses. When an object approaches within preset distances, the glasses vibrate or play an audio cue (bone conduction or buzzer) to warn the wearer.
- **Prototype difficulty:** Beginner → Intermediate (basic soldering + Arduino programming)
- **Estimated cost:** $35–$50

## Circuit & Wiring (short)
- **Power:** LiPo battery → TP4056 charger → 5V USB boost regulator (if using 5V Nano) OR use Arduino that tolerates 3.7V (Pro Mini 3.3V) and skip the boost.
- **HC-SR04:** Vcc to 5V (or 3.3V variants), GND to GND, TRIG to digital pin, ECHO to digital pin.
- **Vibration motor:** Connect motor negative to Arduino GND. Motor positive pins driven by transistor/MOSFET which is controlled by a digital PWM pin on the Arduino. Add a flyback diode across the motor.
- **Bone-conduction / buzzer:** If powered by the Arduino pin, use `tone()` or a small amplifier (LM386 or PAM8302) if louder audio is needed.

> **Safety note:** Don’t drive motors directly from Arduino pins — use a transistor or MOSFET and a diode for protection.

## Firmware (Arduino) — basic example

```cpp
// SmartGlassesBasic.ino
const int trigPin = 9;
const int echoPin = 10;
const int motorPin = 3; // PWM output to MOSFET gate/transistor base via resistor
const int buzzerPin = 5; // optional

void setup() {
  pinMode(trigPin, OUTPUT);
  pinMode(echoPin, INPUT);
  pinMode(motorPin, OUTPUT);
  pinMode(buzzerPin, OUTPUT);
  Serial.begin(9600);
}

long measureDistanceCM() {
  digitalWrite(trigPin, LOW);
  delayMicroseconds(2);
  digitalWrite(trigPin, HIGH);
  delayMicroseconds(10);
  digitalWrite(trigPin, LOW);
  long duration = pulseIn(echoPin, HIGH, 30000); // 30 ms timeout
  if (duration == 0) return 999; // no echo
  long distance = duration / 58; // duration / 58 ≈ cm
  return distance;
}

void loop() {
  long d = measureDistanceCM();
  Serial.println(d);

  if (d < 30) {
    analogWrite(motorPin, 255);
    tone(buzzerPin, 1000);
  } else if (d < 100) {
    noTone(buzzerPin);
    analogWrite(motorPin, 180);
    delay(100);
    analogWrite(motorPin, 0);
    delay(300);
  } else {
    analogWrite(motorPin, 0);
    noTone(buzzerPin);
    delay(50);
  }
}



## Step-by-Step Build Plan (4–6 Weeks)

### **Week 0 — Plan & Buy**
- Finalize parts list and order components.  
- Sketch how to mount the sensor and motor on a pair of glasses you already have.

### **Week 1 — Basic Prototype on Breadboard**
- Wire the **HC-SR04 ultrasonic sensor** to the Arduino and upload a distance-measuring sketch.  
- Print distances to **Serial Monitor** to verify readings.  
- Wire the **vibration motor** to the transistor and test.  
- Program the **three-level feedback logic**.

### **Week 2 — Enclosure & Wearable Mounting**
- Design or make a small bracket for the sensor.  
- Move the motor to the temple area.  
- Solder the final circuit and add **on/off switch** + charger.

### **Week 3 — Testing & Calibration**
- Test with multiple users in different environments.  
- Adjust distance thresholds based on feedback.  
- Measure **battery life**.

### **Week 4 — Polish & Document**
- Tidy up wiring and secure all components.  
- Make a short **demo video**.  
- Publish code and instructions online.

---

## Mounting Tips & Ergonomics
- Keep total weight **under 50 g** on the glasses frame.  
- Use **soft padding** for comfort.  
- Place ultrasonic sensor pointing **straight ahead**.

---

## Testing with Users
- Get **consent** before testing.  
- Start in **controlled environments**.  
- Record feedback and iterate.

---

## Common Problems & Debugging
- **Distance reads 999** → Check wiring and power.  
- **Motor stutters** → Use proper transistor + flyback diode.  
- **Short battery life** → Duty cycle the motor, use a larger battery.  
- **Audio too quiet** → Add an amplifier.

---

## How to Present This on College Apps
- Share your **motivation**.  
- Show your **process with photos**.  
- Include **measurable results**.  
- Open-source your design.

---

## Safety, Limitations & Ethics
- Not a replacement for **professional aids**.  
- Acknowledge **false positives** and detection limits.  
- Respect **user privacy**.

---

## Possible Upgrades
- Multi-sensor for **directionality**.  
- Machine learning for **object recognition**.  
- **Bluetooth logging** + phone app.  
- Pilot study with a **senior center**.

---

## Files to Include in GitHub Repo
- `README.md` with description & photo  
- `BOM.md` or `parts.csv` with costs  
- **Wiring diagram**  
- **Arduino code**  
- Photos & video demos
